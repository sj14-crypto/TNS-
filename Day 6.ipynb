```python
import pandas as pd

data = {
    "Name": ["A", "B", "C"],
    "Joining_Date": ["2022-01-15", "2021-06-20", "2020-12-05"]
}

df = pd.DataFrame(data)

df.info()
# Convert to datetime
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

df

df.info()
```

```python
df["Year"] = df["Joining_Date"].dt.year
df["Month"] = df["Joining_Date"].dt.month
df["Day"] = df["Joining_Date"].dt.day
df["dayname"] = df["Joining_Date"].dt.day_name()

df
```

```python
# Create sample datetime with time
df["Full_DateTime"] = pd.to_datetime([
    "2022-01-15 10:30:00",
    "2021-06-20 14:45:00",
    "2020-12-05 08:15:00"
])

df["Hour"] = df["Full_DateTime"].dt.hour
df["Minute"] = df["Full_DateTime"].dt.minute

df
```

```python
import numpy as np
#Cyclical encoding converts months into circular values so the model understands that December and January are adjacent months.
# Apply on Month
df["Month_sin"] = np.sin(2 * np.pi * df["Month"] / 12)
df["Month_cos"] = np.cos(2 * np.pi * df["Month"] / 12)

df
```

```python
import matplotlib.pyplot as plt

plt.figure()

plt.plot(df["Month"], df["Month_sin"], marker='o', label="sin")
plt.plot(df["Month"], df["Month_cos"], marker='o', label="cos")

plt.xlabel("Month")
plt.ylabel("Value")
plt.title("Cyclical Encoding of Month")
plt.legend()
plt.grid()

plt.show()
```

Step 1: Import libraries and load dataset

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder

df = pd.read_csv('/content/student_data.csv')

print("Original Dataset:")
print(df.head())

print("\nDataset Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())
```

Problem 1: Handling Missing Values

```python
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        df[col] = df[col].fillna(df[col].mean())
    else:
        mode_value = df[col].mode()
        if not mode_value.empty:
            df[col] = df[col].fillna(mode_value[0])

print("Dataset after handling missing values:")
print(df)

print("\nMissing values remaining:")
print(df.isnull().sum())
```

Problem 2: Feature Scaling

```python
print("Before Scaling:")
print(df[['Age', 'Salary']].head())

standard = StandardScaler()
minmax = MinMaxScaler()

df['Salary_Standardized'] = standard.fit_transform(df[['Salary']])
df['Age_Scaled'] = minmax.fit_transform(df[['Age']])

print("\nAfter Scaling:")
print(df[['Age', 'Age_Scaled', 'Salary', 'Salary_Standardized']].head())
```

Problem 3: Encoding Categorical Data

```python
le = LabelEncoder()

df['Gender_Encoded'] = le.fit_transform(df['Gender'].astype(str))

df = pd.get_dummies(df, columns=['City'], dtype=int)

print("Dataset after Encoding:")
print(df.head())
```

Problem 4: Feature Creation

```python
df['Total_Spending'] = df['Salary'] + df['Purchase_Amount']

print("Dataset with Total_Spending:")
print(df[['Salary', 'Purchase_Amount', 'Total_Spending']].head())
```

Problem 5: Date Feature Extraction

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Weekday'] = df['Date'].dt.day_name()

print("Dataset after Date Feature Extraction:")
print(df[['Date', 'Year', 'Month', 'Day', 'Weekday']].head())
```

Problem 6: Binning

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[-np.inf, 25, 45, np.inf],
    labels=['Young', 'Adult', 'Senior']
)

print("Age Groups:")
print(df[['Age', 'Age_Group']].head(10))
```

Problem 7: Outlier Detection using IQR

```python
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Salary'] < lower) | (df['Salary'] > upper)]

print("Lower Limit:", lower)
print("Upper Limit:", upper)

print("\nSalary Outliers:")
print(outliers)
```

Problem 8: Feature Interaction

```python
df['Experience_Salary'] = df['Experience'] * df['Salary']

print("Experience-Salary Feature:")
print(df[['Experience', 'Salary', 'Experience_Salary']].head())
```

Problem 9: Feature Selection

```python
# Select numeric columns
numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation
correlation = numeric_df.corr()['Purchase_Amount'].drop(
    'Purchase_Amount'
)

print("Correlation with Purchase_Amount:")
print(correlation.sort_values())

# Find the least correlated feature
least_important = correlation.abs().idxmin()

print("\nLeast correlated feature:", least_important)

# Drop the feature
df = df.drop(columns=[least_important])

print("\nFinal Dataset:")
print(df.head())

print("\nFinal Dataset Shape:", df.shape)
```
